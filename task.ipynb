```python
# Previous code block(s)
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

data = {
"Customer ID": [101, 102, 103, 101, 104],
"Product Name": ["Widget A", "Widget B", "Widget C", "Widget A", "Widget D"],
"Category": ["Widgets", "Widgets", "Tools", "Widgets", "Tools"],
"Quantity Purchased": [5, 3, 2, 7, 1],
"Price per Unit": [20, 30, 40, 20, 50],
"Total Spent": [100, 90, 80, 140, 50],
"Purchase Date": ["2025-03-21", "2025-03-22", "2025-03-23", "2025-03-24", "2025-03-25"]
}
df = pd.DataFrame(data)
df["Purchase Date"] = pd.to_datetime(df["Purchase Date"])
print(df)


# %%
print(df.isnull().sum())
# %%
print(df.duplicated().sum())
# %%
print(df.describe())
# %%
df['day of week'] = df['Purchase Date'].dt.day_name()
print(df.head())
# %%
df['total sales'] = df['Quantity Purchased'] * df['Price per Unit'] # Create 'total sales' column
print(df.head())
# %%
daily_sales = df.groupby('Purchase Date')['total sales'].sum()
plt.plot(daily_sales.index, daily_sales.values)
plt.xlabel('Purchase Date')
plt.ylabel('Total Sales')
plt.title('Daily Sales')
plt.show()
# %%
# Now, 'total sales' column should exist
top_products = df.groupby('Product Name')['total sales'].sum().nlargest(5)
top_products.plot(kind='bar')
plt.xlabel('Product Name')
plt.ylabel('Total Sales')
plt.title('Top Products')
plt.show()
```

```python
customer_spending = df.groupby('Customer ID')['Total Spent'].sum()
plt.hist(customer_spending, bins=20)
plt.xlabel('Total Spent')
plt.ylabel('Frequency')
plt.title('Customer Spending Distribution')
plt.show()
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Assuming df is already defined from previous cells
# If not, you need to define df here as well

# Define df if it's not already defined
data = {
    "Customer ID": [101, 102, 103, 101, 104],
    "Product Name": ["Widget A", "Widget B", "Widget C", "Widget A", "Widget D"],
    "Category": ["Widgets", "Widgets", "Tools", "Widgets", "Tools"],
    "Quantity Purchased": [5, 3, 2, 7, 1],
    "Price per Unit": [20, 30, 40, 20, 50],
    "Total Spent": [100, 90, 80, 140, 50],
    "Purchase Date": ["2025-03-21", "2025-03-22", "2025-03-23", "2025-03-24", "2025-03-25"]
}
df = pd.DataFrame(data)
df["Purchase Date"] = pd.to_datetime(df["Purchase Date"])
df['total sales'] = df['Quantity Purchased'] * df['Price per Unit'] # Calculate 'total sales'

# Recalculate daily_sales to make it available in this cell
daily_sales = df.groupby('Purchase Date')['total sales'].sum()

# Now you can use daily_sales in your plot
sns.lineplot(x=daily_sales.index, y=daily_sales.values)
plt.xlabel('Purchase Date')
plt.ylabel('Total Sales')
plt.title('Sales over time')
plt.show()
```

```python
sns.barplot(x=daily_sales.index, y=daily_sales.values)
plt.xlabel('Purchase Date')
plt.ylabel('Total Sales')
plt.title('Daily Sales')
plt.show()
```
